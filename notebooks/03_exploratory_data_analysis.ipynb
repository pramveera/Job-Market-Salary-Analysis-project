## Load the processed data

```python
import pandas as pd
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent
PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"

df = pd.read_csv(
    PROCESSED_DIR / "job_postings_flagged.csv"
)

print("Dataset shape:", df.shape)
df.head()
```

### Step 2 — Find the most common job titles

```python
top_jobs = df["title"].value_counts().head(10)

print(top_jobs)
```

### Step 3 — Find the most common locations

```python
top_locations = df["location"].value_counts().head(10)

print(top_locations)
```

### Step 4 — Analyze experience levels

```python
experience_counts = df[
    "formatted_experience_level"
].value_counts()

print(experience_counts)
```

```python
import matplotlib.pyplot as plt
```

Chart 1 — Top 10 job titles

```python
top_jobs = df["title"].value_counts().head(10)

top_jobs.sort_values().plot(
    kind="barh",
    figsize=(10, 6)
)

plt.title("Top 10 Most Frequently Posted Job Titles")
plt.xlabel("Number of Job Postings")
plt.ylabel("Job Title")
plt.tight_layout()
plt.show()
```

Chart 2 — Experience-level distribution

```python
experience_counts = df[
    "formatted_experience_level"
].value_counts()

experience_counts.plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Job Postings by Experience Level")
plt.xlabel("Experience Level")
plt.ylabel("Number of Postings")
plt.xticks(rotation=35, ha="right")
plt.tight_layout()
plt.show()
```

Chart 3 — Top 10 listed locations

```python
top_locations = df["location"].value_counts().head(10)

top_locations.sort_values().plot(
    kind="barh",
    figsize=(10, 6)
)

plt.title("Top 10 Locations by Job Postings")
plt.xlabel("Number of Job Postings")
plt.ylabel("Location")
plt.tight_layout()
plt.show()
```
